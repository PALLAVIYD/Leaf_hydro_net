```python
# ==========================================================
# 1️⃣ Install Required Libraries
# ==========================================================
!pip install tensorflow tensorflow-datasets matplotlib seaborn scikit-learn -q

# ==========================================================
# 2️⃣ Import Libraries
# ==========================================================
import tensorflow as tf
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

# ==========================================================
# 3️⃣ Load PlantVillage Dataset (Auto Download)
# ==========================================================
(ds_train, ds_test), ds_info = tfds.load(
    'plant_village',
    split=['train[:80%]', 'train[80%:]'],
    as_supervised=True,
    with_info=True
)

NUM_CLASSES = ds_info.features['label'].num_classes
IMG_SIZE = 128
BATCH_SIZE = 32

print("Number of Classes:", NUM_CLASSES)

# ==========================================================
# 4️⃣ Preprocess Dataset
# ==========================================================
def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = image / 255.0
    return image, label

ds_train = ds_train.map(preprocess).shuffle(1000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
ds_test = ds_test.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# ==========================================================
# 5️⃣ Build CNN Model
# ==========================================================
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(IMG_SIZE, IMG_SIZE, 3)),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(128, (3,3), activation='relu'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dense(NUM_CLASSES, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ==========================================================
# 6️⃣ Train Model
# ==========================================================
history = model.fit(ds_train, validation_data=ds_test, epochs=5)

# ==========================================================
# 7️⃣ Evaluate Model
# ==========================================================
test_loss, test_acc = model.evaluate(ds_test)
print("\nTest Accuracy:", round(test_acc*100,2), "%")

# ==========================================================
# 8️⃣ Show Sample Predictions
# ==========================================================
class_names = ds_info.features['label'].names

for images, labels in ds_test.take(1):
    predictions = model.predict(images)
    preds = np.argmax(predictions, axis=1)

plt.figure(figsize=(12,6))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i])
    plt.title(f"True: {class_names[int(labels[i])]}\nPred: {class_names[int(preds[i])]}")
    plt.axis("off")
plt.show()

# ==========================================================
# 9️⃣ Confusion Matrix
# ==========================================================
y_true = []
y_pred = []

for images, labels in ds_test:
    preds = model.predict(images)
    preds = np.argmax(preds, axis=1)
    y_true.extend(labels.numpy())
    y_pred.extend(preds)

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8,6))
sns.heatmap(cm, cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.show()

# ==========================================================
# 🔟 Training Curve
# ==========================================================
plt.figure(figsize=(8,4))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.legend()
plt.title("Training Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.show()
```

```python
IMG_SIZE = 128

def preprocess(image, label):
    image = tf.image.resize(image,(IMG_SIZE,IMG_SIZE))
    image = image/255.0
    return image,label

ds_train = ds_train.map(preprocess)
ds_test = ds_test.map(preprocess)
```

```python
images = []
labels = []

for img,label in ds_train.take(50):
    images.append(img.numpy())
    labels.append(label.numpy())

print("Images Loaded:",len(images))
```

```python
import cv2
import numpy as np

img = (images[0]*255).astype(np.uint8)

edges = cv2.Canny(img,50,150)

plt.imshow(edges,cmap="gray")
plt.title("Leaf Edge Detection")
plt.axis("off")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = (images[0]*255).astype(np.uint8)

plt.imshow(img)
plt.title("Original Leaf")
plt.axis("off")
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)

lower_green = np.array([25,40,40])
upper_green = np.array([90,255,255])

mask = cv2.inRange(hsv, lower_green, upper_green)

plt.imshow(mask, cmap="gray")
plt.title("Leaf Mask")
plt.axis("off")
```

```python
contours,_ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

largest = max(contours, key=cv2.contourArea)

leaf_mask = np.zeros_like(mask)
cv2.drawContours(leaf_mask,[largest],-1,255,-1)

plt.imshow(leaf_mask,cmap="gray")
plt.title("Segmented Leaf")
plt.axis("off")
```

```python
x_coords = largest[:,:,0]

current_width = x_coords.max() - x_coords.min()

print("Current Width:", current_width)
```

```python
baseline_width = 200

LRI = (baseline_width - current_width)/baseline_width

print("Leaf Rolling Index:", round(LRI,3))
```

```python
if LRI < 0.1:
    stage = "Healthy"

elif LRI < 0.3:
    stage = "Early Drought"

elif LRI < 0.5:
    stage = "Moderate Stress"

else:
    stage = "Severe Drought"

print("Predicted Stage:", stage)
```

```python
plt.figure(figsize=(6,6))

cv2.drawContours(img,[largest],-1,(255,0,0),3)

plt.imshow(img)
plt.title(f"LRI: {round(LRI,2)} | {stage}")
plt.axis("off")
```

```python
results = []

for image in images[:20]:

    img = (image*255).astype(np.uint8)

    hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)
    mask = cv2.inRange(hsv, lower_green, upper_green)

    contours,_ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    if len(contours)==0:
        continue

    largest = max(contours,key=cv2.contourArea)

    x = largest[:,:,0]
    width = x.max()-x.min()

    LRI = (200-width)/200

    results.append(LRI)

print("Average LRI:", np.mean(results))
```
